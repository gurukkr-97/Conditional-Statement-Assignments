## Even/Odd Number Counter

## Reverse a Number


```python
user_num = int(input('Enter Your Number'))
rev_num = 0
while user_num>0:
    last_num = user_num % 10
    rev_num = rev_num*10 + last_num
    user_num //= 10
print(f"The Reverse of your Number is {rev_num}")
```

## Palindrom

```python
user_num = int(input('Enter Your Number'))
c_usernum = user_num
rev_num = 0
while user_num>0:
    last_num = user_num % 10
    rev_num = rev_num*10 + last_num
    user_num //= 10
if rev_num == c_usernum:
    print(f"The given number {c_usernum} is a Palindrome")
else:
    print(f"The given number {c_usernum} is not a Palindrome")
```

## Prime Number Check

```python
num = int(input('Enter Your Number'))
sqrtnum = int(num**0.5)
count = 0
for i in range (1, sqrtnum+1):
    if num % i== 0:
        count += 1
if count == 1:
    print(f"Your Number {num} is Prime Number")
else :
    print (f"Your Number {num} is not Prime Number")
```

## Factorial of a number

```python
num = int(input('Enter Your Number'))
finalresult = 0
temp = 1
count = 1
while count <= num:
    fact = temp * count
    count += 1
    temp = fact
finalresult = fact 
print(f"The factorial of your num {num} is {finalresult}")
```

## Fibonacci series

```python
num = int(input("Enter your number "))
num1 = 0
num2 = 1
fibo = [num1,num2]
count = 3
while count <= num:
    num3 = num1 + num2
    fibo.append(num3)
    num1 = num2
    num2 = num3
    count += 1
print(f"The Fibanocci series of {num} terms are : {fibo}")
```

```python
num = int(input("Enter your number "))
num1 = 0
num2 = 1
fibo = [num1,num2]
for i in range (num-2):
    i = num1 + num2
    fibo.append(i)
    num1 = num2
    num2 = i
print(f"The Fibanocci series of {num} terms are : {fibo}")
```
